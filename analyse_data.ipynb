```python
import pandas as pd
```

```python
df = pd.read_csv("flood_pics_with_metadata.csv")
```

```python
df.head()
```

```python
recorded_dates = df["recorded_date"]
```

```python
#convert YYY-MM-DDTHH:MM:SS to year
df["year"] = pd.to_datetime(df["recorded_date"]).dt.year
```

```python
df.tail()
```

```python
# count the number of images per year
images_per_year = df.groupby(df['year']).size()
```

```python
images_per_year.plot(kind="bar", width=0.9, figsize=(15, 10), title="Number of Flood Images per Year", xlabel="Year", ylabel="Number of Images")
```

```python
# select all rows after 2008 
df_after_2008 = df[df["year"] > 2008]
```

```python
import geopandas as gpd
```

```python
gdf = gpd.read_file("/Users/chamodya/Workspace/PhD/phd_repo/statsnz-regional-council-2025-clipped-SHP")
```

```python
gdf
```

```python
# convert coordinates from df_after_2008 to EPSG:2193 and count the number of coordinates within each region in the gdf
from shapely.geometry import Point
geometry = [Point(xy) for xy in zip(df_after_2008["longitude"], df_after_2008["latitude"])]
gdf_after_2008 = gpd.GeoDataFrame(df_after_2008, geometry=geometry, crs="EPSG:4326")
gdf_after_2008 = gdf_after_2008.to_crs(epsg=2193)
```

```python
# if gdf_after_2008.geometry is contained inside gdf.geometry, then assign the region name to a new column in gdf_after_2008
gdf_after_2008["region"] = gdf_after_2008.geometry.apply(lambda x: gdf[gdf.geometry.contains(x)]["REGC2025_2"].values[0] if gdf.geometry.contains(x).any() else "Unknown")
```

```python
gdf_after_2008.groupby("region").size().plot(kind="bar", width=0.9, figsize=(15, 10), title="Number of Flood Images per Region (after 2008)", xlabel="Region", ylabel="Number of Images")
```

```python
df_after_2008.to_csv("flood_pics_after_2008.csv", index=False)
```
