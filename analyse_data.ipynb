```python
def get_gdf_after_year(df, year, gdf):
    df_after_year = df[df["year"] >= year]
    geometry = [Point(xy) for xy in zip(df_after_year["longitude"], df_after_year["latitude"])]
    # if gdf_after_2008.geometry is contained inside gdf.geometry, then assign the region name to a new column in gdf_after_2008
    gdf_after_year = gpd.GeoDataFrame(df_after_year, geometry=geometry, crs="EPSG:4326")
    gdf_after_year = gdf_after_year.to_crs(epsg=2193)
    gdf_after_year["region"] = gdf_after_year.geometry.apply(lambda x: gdf[gdf.geometry.contains(x)]["REGC2025_2"].values[0] if gdf.geometry.contains(x).any() else "Unknown")
    return gdf_after_year

def get_gdf_on_year(df, year, gdf):
    df_on_year = df[df["year"] == year]
    geometry = [Point(xy) for xy in zip(df_on_year["longitude"], df_on_year["latitude"])]
    # if gdf_after_2008.geometry is contained inside gdf.geometry, then assign the region name to a new column in gdf_after_2008
    gdf_on_year = gpd.GeoDataFrame(df_on_year, geometry=geometry, crs="EPSG:4326")
    gdf_on_year = gdf_on_year.to_crs(epsg=2193)
    gdf_on_year["region"] = gdf_on_year.geometry.apply(lambda x: gdf[gdf.geometry.contains(x)]["REGC2025_2"].values[0] if gdf.geometry.contains(x).any() else "Unknown")
    return gdf_on_year
```

```python
def visualize_flood_images_per_region(df, gdf, title, year):
    gdf_after_year = get_gdf_after_year(df, year, gdf)
    return gdf_after_year.groupby("region").size().plot(kind="bar", width=0.9, figsize=(15, 10), title=title, xlabel="Region", ylabel="Number of Images")
```

```python
# fill unknown regions based on the region it belongs or closest to based on latitude and longitude in the gdf
def fill_unknown_regions(test_gdf, gdf):
    for index, row in test_gdf.iterrows():
        if row["region"] == "Unknown":
            point = row["geometry"]
            region = gdf[gdf.geometry.contains(point)]["REGC2025_2"].values[0] if gdf.geometry.contains(point).any() else "Unknown"
            if region == "Unknown":
                # assign the closes region to the point
                region = gdf.iloc[gdf.distance(point).sort_values().index[0]].REGC2025_2
            test_gdf.at[index, "region"] = region
    return test_gdf
    
```

```python
# function to visualize the yearly distribution of flood images on a map of New Zealand, with the option to filter by year
# the points in each year should be colored differently
def visualize_flood_images_on_map_by_year(df, gdf, years=range(2017,2026)):
    fig, ax = plt.subplots(figsize=(15, 10))
    gdf.plot(ax=ax, color="lightgrey", edgecolor="black")
    color_range = plt.cm.get_cmap("tab10", len(years))
    for i, year in enumerate(years):
        gdf_after_year = get_gdf_on_year(df, year, gdf)
        gdf_after_year.plot(ax=ax, color=color_range(i), markersize=5, label=str(year))
    plt.legend(years, title="Year", loc="upper right")
    plt.title("Flood Images (after {}) on Map of New Zealand".format(gdf_after_year["year"].min()))
    plt.xlabel("Longitude")
    plt.ylabel("Latitude")
    plt.show()
```
